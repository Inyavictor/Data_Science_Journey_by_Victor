DATA-VISUALIZING-QUANTITIES

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data_visualization = pd.read_csv("../data/honey.csv")
data_visualization
```

```python
plt.bar (data_visualization['yieldpercol'], data_visualization['totalprod'], color='green')
plt.xlabel('Yield per Colony')
plt.ylabel('Total Production')
plt.title('Honey Production by Yield per Colony')
plt.figure(figsize=(10, 6))
plt.show()
```

```python
plt.plot(data_visualization['year'], data_visualization['prodvalue'], color='blue', marker='o', linestyle='-')
plt.xlabel('Year')
plt.ylabel('Production Value')
plt.title('Honey Production Over Time')
plt.grid(True)
plt.show()
```

```python
plt.bar(data_visualization['state'], data_visualization['totalprod'], color='orange')
plt.xlabel('State')
plt.ylabel('Total Production')
plt.title('Honey Production by State')
plt.xticks(rotation=90)
plt.figure(figsize=(10, 20))
plt.show()
```

```python
plt.hist( data_visualization['yieldpercol'], bins=20, color='lightcoral', edgecolor='black')
plt.xlabel('Yield per Colony')
plt.ylabel('Frequency')
plt.title('Distribution of Honey Production')
plt.show()
```
